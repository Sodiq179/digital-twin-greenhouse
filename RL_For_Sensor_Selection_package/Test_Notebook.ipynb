```python

print("========== 1. TESTING PSYCHROMETRIC CALCULATIONS ==========")
# Given standard conditions: 25°C and 50% Relative Humidity [cite: 147]
T_C, RH = 25.0, 50.0

# Run the function 
dp, W, h, sv = compute_psychrometrics(T_C, RH)

print(f"Input Conditions  -> Temperature: {T_C}°C | Relative Humidity: {RH}%")
print(f"Dew Point         -> {dp:.2f} °C")
print(f"Humidity Ratio    -> {W:.5f} kg/kg")
print(f"Enthalpy          -> {h:.2f} kJ/kg")
print(f"Specific Volume   -> {sv:.4f} m³/kg\n")


print("========== 2. TESTING THOMPSON SAMPLING ==========")
# Create dummy dataframe mimicking sensor data across 5 timestamps 
data = {
    'A1': [20.5, 22.1, 19.8, 21.0, 23.2],
    'A2': [22.0, 24.5, 21.1, 25.0, 26.3],
    'A3': [18.0, 19.2, 18.5, 20.1, 21.0]
}
df_sensors = pd.DataFrame(data)

print("--- Mock Sensor Data ---")
print(df_sensors, "\n")

# Run the algorithm 
top_sensors, counts = thompson_sampling_mean_crossing(df_sensors)

print("--- Algorithm Outputs ---")
print(f"Top Sensors (Ranked by Index): {top_sensors}")
print(f"Total Selection Counts: {counts}\n")


print("========== 3. TESTING DATA PREPARATION & MERGING ==========")
# Dummy sensor coordinates 
coords_data = {
    'Sensor': ['A1', 'A2'], 
    'X': [0, 10], 
    'Y': [0, 10], 
    'Z': [3, 3]
}
sensor_coords = pd.DataFrame(coords_data)

# Dummy environmental data mimicking the Excel layout 
env_data = {
    'Date/ Time': ['2020-04-01 12:00', '2020-04-01 12:10'],
    'A1. Temp. (°C)': [25.0, 26.0],
    'A1. Humidity (%)': [50.0, 55.0],
    'A2. Temp. (°C)': [22.0, 23.0],
    'A2. Humidity (%)': [40.0, 45.0]
}
df_env = pd.DataFrame(env_data)

# Run the function 
merged_df = prepare_psychro_merged(df_env, sensor_coords)

print("--- Merged Resulting DataFrame ---")
display(merged_df) # Using display() renders a nicely formatted table in Jupyter Notebooks
```
